Q12: What are the basic steps involved in exception handling in Python?
   - The basic steps involved in exception handling in Python revolve around the try, except, else, and finally blocks:
     - try block: This block encloses the code that is anticipated to potentially raise an exception. Python will attempt to execute the code within this block. If an exception occurs, the execution of the try block is halted, and control is passed to the appropriate except block.
     - except block(s): Following a try block, one or more except blocks can be used to handle specific types of exceptions. If an exception of the specified type (or a subclass of it) occurs within the try block, the code within the corresponding except block is executed. You can have multiple except blocks to handle different exceptions differently, or a general except block to catch any unhandled exception.
     - else block (optional): An else block can be included after all except blocks. The code within the else block is executed only if no exception occurred in the try block. This is useful for code that should only run when the try block completes successfully.
     - finally block (optional): A finally block can be used after all try, except, and else blocks. The code within the finally block is always executed, regardless of whether an exception occurred in the try block or was handled by an except block. This block is typically used for cleanup operations, such as closing files or releasing resources, ensuring they are performed even if an error arises.

     - Monitors for exceptions: When the code within the try block executes, the system monitors for any exceptions that might be raised. If an exception occurs, the execution of the try block is immediately halted.

   - Role of except:

It provides a mechanism to catch and handle specific types of exceptions. You can specify the type of exception to catch (e.g., ZeroDivisionError, ValueError, FileNotFoundError), allowing for tailored responses to different error scenarios.
     - Provides recovery or alternative actions: Within the except block, you can implement code to recover from the error, log the error for debugging, display a user-friendly message, or take alternative actions to ensure the program continues to function, albeit in a modified way.
     - Prevents program termination: By handling the exception, the except block prevents the program from crashing and terminating abruptly, leading to a more robust and user-friendly application.

```python
# Q2: Write a Python program to read the contents of a file and print each line.

def read_and_print_file(filename):
    """
    Reads the contents of a specified file and prints each line.

    Args:
        filename (str): The path to the file to be read.
    """
    try:
        with open(filename, 'r') as file:
            for line in file:
                print(line.strip())  # .strip() removes leading/trailing whitespace, including newlines
    except FileNotFoundError:
        print(f"Error: The file '{filename}' was not found.")
    except Exception as e:
        print(f"An unexpected error occurred: {e}")

# Example usage:
if __name__ == "__main__":
    # Create a dummy file for testing
    with open("sample.txt", "w") as f:
        f.write("This is the first line.\n")
        f.write("This is the second line.\n")
        f.write("And this is the third line.")

    file_to_read = "sample.txt"
    read_and_print_file(file_to_read)

    # Example with a non-existent file
    read_and_print_file("non_existent_file.txt")
```

```python
#Q4: Write a Python script that reads from one file and writes its content to another file.

def copy_file_content(source_file_path, destination_file_path):
    """
    Reads the content of a source file and writes it to a destination file.

    Args:
        source_file_path (str): The path to the file to read from.
        destination_file_path (str): The path to the file to write to.
    """
    try:
        with open(source_file_path, 'r') as source_file:
            content = source_file.read()

        with open(destination_file_path, 'w') as destination_file:
            destination_file.write(content)

        print(f"Content successfully copied from '{source_file_path}' to '{destination_file_path}'.")

    except FileNotFoundError:
        print(f"Error: One of the files was not found. Please check the paths.")
    except Exception as e:
        print(f"An error occurred: {e}")

if __name__ == "__main__":
    # Example usage:
    # Create a dummy source file for demonstration
    with open("source.txt", "w") as f:
        f.write("This is the content of the source file.\n")
        f.write("It has multiple lines.\n")
        f.write("Hello, world!")

    # Define the paths for the source and destination files
    input_file = "source.txt"
    output_file = "destination.txt"

    # Call the function to copy the content
    copy_file_content(input_file, output_file)

    # Verify the content of the destination file (optional)
    try:
        with open(output_file, 'r') as f:
            print(f"\nContent of '{output_file}':")
            print(f.read())
    except FileNotFoundError:
        print(f"Error: Destination file '{output_file}' not found for verification.")
```

```python
#Q11: Write a Python program that uses a try-except block to handle an error when attempting to access a dictionary key that doesn't exist.

def get_value_from_dict(dictionary, key):
    """
    Attempts to retrieve a value from a dictionary using a given key.
    Handles KeyError if the key is not found.
    """
    try:
        value = dictionary[key]
        print(f"The value for '{key}' is: {value}")
    except KeyError:
        print(f"Error: The key '{key}' does not exist in the dictionary.")
    except Exception as e: # Catch any other unexpected exceptions
        print(f"An unexpected error occurred: {e}")

# Example usage
my_dict = {
    "name": "Alice",
    "age": 30,
    "city": "New York"
}

# Attempt to access an existing key
get_value_from_dict(my_dict, "name")

# Attempt to access a non-existent key
get_value_from_dict(my_dict, "occupation")

# Another attempt with a non-existent key
get_value_from_dict(my_dict, "email")
```

```python
#Q14: Write a program that uses the logging module to log both informational and error messages.

import logging

def setup_logging():
    """
    Configures the logging system to log informational messages to 'info.log',
    error messages to 'error.log', and all messages (INFO and above) to the console.
    """
    # Create a custom logger
    logger = logging.getLogger(__name__)
    logger.setLevel(logging.INFO) # Set the base level for the logger

    # Create handlers
    console_handler = logging.StreamHandler()
    info_file_handler = logging.FileHandler('info.log')
    error_file_handler = logging.FileHandler('error.log')

    # Set levels for handlers
    console_handler.setLevel(logging.INFO)
    info_file_handler.setLevel(logging.INFO)
    error_file_handler.setLevel(logging.ERROR)

    # Create formatters and add them to handlers
    formatter = logging.Formatter('%(asctime)s - %(name)s - %(levelname)s - %(message)s')
    console_handler.setFormatter(formatter)
    info_file_handler.setFormatter(formatter)
    error_file_handler.setFormatter(formatter)

    # Add handlers to the logger
    logger.addHandler(console_handler)
    logger.addHandler(info_file_handler)
    logger.addHandler(error_file_handler)

    return logger

def main():
    """
    Main function to demonstrate logging of informational and error messages.
    """
    logger = setup_logging()

    logger.info("This is an informational message.")
    logger.warning("This is a warning message.") # Will be logged to console and info.log
    logger.error("This is an error message. Something went wrong!") # Will be logged to all handlers
    logger.debug("This is a debug message.") # Will not be logged due to logger's level

    try:
        result = 10 / 0
    except ZeroDivisionError as e:
        logger.exception("An exception occurred during division.") # Logs error with traceback

if __name__ == "__main__":
    main()
```

```python
#Q16: Demonstrate how to use memory profiling to check the memory usage of a small program.

from memory_profiler import profile
import requests

class BaseExtractor:
    # Function to write words from a list into a file
    @profile  # This decorator will monitor memory usage
    def parse_list(self, array):
        with open('words.txt', 'w') as f:
            for word in array:
                f.writelines(word + "\n")

    # Function to fetch data from a URL and save it
    @profile
    def parse_url(self, url):
        response = requests.get(url).text
        with open('url.txt', 'w') as f:
            f.writelines(response)
```

```python
#Q17: Write a Python program to create and write a list of numbers to a file, one number per line.

def write_numbers_to_file(numbers_list, filename):
    """
    Writes a list of numbers to a specified file, with each number on a new line.

    Args:
        numbers_list (list): A list of numbers (integers or floats).
        filename (str): The name of the file to write to.
    """
    try:
        with open(filename, 'w') as file:
            for number in numbers_list:
                file.write(str(number) + '\n')
        print(f"Numbers successfully written to '{filename}'.")
    except IOError as e:
        print(f"Error writing to file '{filename}': {e}")

# Example usage:
if __name__ == "__main__":
    my_numbers = [10, 25, 30, 45, 50.5, 60]
    output_file = "my_numbers.txt"

    write_numbers_to_file(my_numbers, output_file)

    # You can also create a list using range()
    more_numbers = list(range(1, 11)) # Numbers from 1 to 10
    another_output_file = "range_numbers.txt"
    write_numbers_to_file(more_numbers, another_output_file)
```

```python
#Q18: How would you implement a basic logging setup that logs to a file with rotation after 1MB?

import logging
from logging.handlers import RotatingFileHandler

# Define the log file name, maximum size (1MB), and number of backup files
LOG_FILE = 'application.log'
MAX_BYTES = 1 * 1024 * 1024  # 1 MB
BACKUP_COUNT = 5  # Keep 5 backup log files

# Create a logger instance
logger = logging.getLogger(__name__)
logger.setLevel(logging.INFO)  # Set the logging level

# Create a RotatingFileHandler
# This handler rotates the log file when it reaches MAX_BYTES,
# and keeps BACKUP_COUNT older log files.
handler = RotatingFileHandler(LOG_FILE, maxBytes=MAX_BYTES, backupCount=BACKUP_COUNT)

# Define a formatter for the log messages
formatter = logging.Formatter('%(asctime)s - %(name)s - %(levelname)s - %(message)s')

# Set the formatter for the handler
handler.setFormatter(formatter)

# Add the handler to the logger
logger.addHandler(handler)

# Example usage:
logger.info("This is an informational message.")
logger.warning("This is a warning message.")
logger.error("This is an error message.")
```

```python
#Q19: Write a program that handles both IndexError and KeyError using a try-except block.

def handle_lookup_errors(data, key_or_index):
    """
    Attempts to access an element from a list or dictionary, handling
    IndexError for lists and KeyError for dictionaries.

    Args:
        data: A list or a dictionary.
        key_or_index: The key (for dictionaries) or index (for lists) to access.
    """
    try:
        if isinstance(data, list):
            value = data[key_or_index]
            print(f"Accessed list element at index {key_or_index}: {value}")
        elif isinstance(data, dict):
            value = data[key_or_index]
            print(f"Accessed dictionary value with key '{key_or_index}': {value}")
        else:
            print("Error: 'data' must be a list or a dictionary.")
    except IndexError:
        print(f"Error: IndexError - Index {key_or_index} is out of range for the list.")
    except KeyError:
        print(f"Error: KeyError - Key '{key_or_index}' not found in the dictionary.")
    except TypeError:
        print("Error: TypeError - Invalid key or index type for the given data structure.")
    except Exception as e:
        print(f"An unexpected error occurred: {e}")

# Example usage with a list
my_list = [10, 20, 30]
handle_lookup_errors(my_list, 1)  # Valid index
handle_lookup_errors(my_list, 5)  # IndexError

# Example usage with a dictionary
my_dict = {"apple": 1, "banana": 2}
handle_lookup_errors(my_dict, "apple")  # Valid key
handle_lookup_errors(my_dict, "orange") # KeyError

# Example with incorrect data type
handle_lookup_errors("a string", 0) # TypeError, caught by generic Exception

# Example with an unexpected error (e.g., trying to use a string as an index on a list)
handle_lookup_errors(my_list, "hello")
```

```python
#Q21: Write a Python program that reads a file and prints the number of occurrences of a specific word.

def count_word_occurrences(filename, target_word):
    """
    Counts the occurrences of a specific word in a text file.

    Args:
        filename (str): The path to the text file.
        target_word (str): The word to search for.

    Returns:
        int: The number of occurrences of the target word.
    """
    count = 0
    try:
        with open(filename, 'r', encoding='utf-8') as file:
            for line in file:
                # Convert line to lowercase to ensure case-insensitive matching
                # and split into words.
                words = line.lower().split()
                # Count occurrences of the target word in the current line
                count += words.count(target_word.lower())
        return count
    except FileNotFoundError:
        print(f"Error: File '{filename}' not found.")
        return -1
    except Exception as e:
        print(f"An error occurred: {e}")
        return -1

if __name__ == "__main__":
    file_name = input("Enter the filename (e.g., my_text_file.txt): ")
    word_to_find = input("Enter the word to count: ")

    occurrences = count_word_occurrences(file_name, word_to_find)

    if occurrences != -1:
        print(f"The word '{word_to_find}' appears {occurrences} times in '{file_name}'.")
```

```python
#Q22: How can you check if a file is empty before attempting to read its contents?

import os

file_path = "your_file.txt"

# Method 1: Using os.path.getsize()
if os.path.exists(file_path):
    if os.path.getsize(file_path) == 0:
        print(f"The file '{file_path}' is empty.")
    else:
        print(f"The file '{file_path}' is not empty.")
else:
    print(f"The file '{file_path}' does not exist.")

# Method 2: Using file.seek() and file.tell()
if os.path.exists(file_path):
    with open(file_path, 'r') as file_obj:
        file_obj.seek(0, os.SEEK_END)  # Move pointer to the end
        if file_obj.tell() == 0:  # Get current position (file size)
            print(f"The file '{file_path}' is empty.")
        else:
            print(f"The file '{file_path}' is not empty.")
else:
    print(f"The file '{file_path}' does not exist.")
```

```python
#Q23: Write a Python program that writes to a log file when an error occurs during file handling.

import logging
import os

# Configure the logger
log_file_name = "file_handling_errors.log"
logging.basicConfig(
    filename=log_file_name,
    level=logging.ERROR,  # Only log messages of level ERROR and above
    format='%(asctime)s - %(levelname)s - %(message)s'
)

def process_file(file_path):
    """
    Attempts to open and read a file, logging any errors encountered.
    """
    try:
        with open(file_path, 'r') as f:
            content = f.read()
            print(f"Successfully read content from {file_path}:")
            print(content[:50] + "..." if len(content) > 50 else content)
    except FileNotFoundError:
        logging.error(f"Error: File not found at '{file_path}'")
    except PermissionError:
        logging.error(f"Error: Permission denied when accessing '{file_path}'")
    except IOError as e:
        logging.error(f"Error: An I/O error occurred with '{file_path}': {e}")
    except Exception as e:
        logging.error(f"An unexpected error occurred while processing '{file_path}': {e}")

if __name__ == "__main__":
    # Create a dummy file for successful operation
    with open("successful_file.txt", "w") as f:
        f.write("This is a test file for successful reading.")

    # Test cases
    process_file("successful_file.txt")  # Should succeed
    process_file("non_existent_file.txt") # Should trigger FileNotFoundError

    # Simulate a permission error (may require specific OS setup to truly test)
    # On Linux/macOS, you might create a file and remove read permissions:
    # os.chmod("no_permission.txt", 0o222) # Write-only permissions
    # process_file("no_permission.txt")

    # Clean up dummy file
    os.remove("successful_file.txt")

    print(f"\nCheck '{log_file_name}' for any logged errors.")
```
